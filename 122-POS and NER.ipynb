```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import spacy
import re
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bbc_data=pd.read_csv('bbc_news.csv')
```

```python
bbc_data.head(6)
```

```python
bbc_data.info()
```

```python
titels=pd.DataFrame(bbc_data['title'])
titels.head()
```

```python
titels['Lower_Case']=titels['title'].str.lower()   #.str tells panda to apply .lower() on every row.
titels.head()
```

```python
en_stopwords=stopwords.words('english')
titels['No_stopwords']=titels['Lower_Case'].apply(lambda x: ' '.join([word for word in x.split() if word not in (en_stopwords)]))
titels.head()
```

```python
# punctation removal
titels['no_stopwords_no_punct'] = titels.apply(lambda x: re.sub(r"([^\w\s])", "", x['No_stopwords']), axis=1)
titels.head()
```

```python

# tokenize
titels['tokens_raw'] = titels.apply(lambda x: word_tokenize(x['title']), axis=1)
titels['tokens_clean'] = titels.apply(lambda x: word_tokenize(x['no_stopwords_no_punct']), axis=1)
titels.head()
```

```python
# lemmatizing 
lemmatizer = WordNetLemmatizer()
titels["tokens_clean_lemmatized"] = titels["tokens_clean"].apply(lambda tokens: [lemmatizer.lemmatize(token) for token in tokens])
titels.head()
```

```python
# create lists for just our tokens

tokens_raw_list = sum(titels['tokens_raw'], []) #unpack our lists into a single list
tokens_clean_list = sum(titels['tokens_clean_lemmatized'], [])
```

```python
nlp = spacy.load('en_core_web_sm')
# create a spacy doc from our raw text - better for pos tagging
spacy_doc = nlp(' '.join(tokens_raw_list))
```

```python
pos_df = pd.DataFrame(columns=['token', 'pos_tag'])

for token in spacy_doc:
    pos_df = pd.concat([pos_df,
                       pd.DataFrame.from_records([{'token': token.text,'pos_tag': token.pos_}])], ignore_index=True)
```

```python
# token frequency count
pos_df_counts = pos_df.groupby(['token','pos_tag']).size().reset_index(name='counts').sort_values(by='counts', ascending=False)
pos_df_counts.head(10)
```

```python
pos_df_counts[:7].sort_values(by='counts').plot.bar(color='blue',width=.5,figsize=(12,8))
```

```python
# most common nouns
nouns = pos_df_counts[pos_df_counts.pos_tag == "NOUN"][0:10]
nouns
```

```python
# most common verbs
verbs = pos_df_counts[pos_df_counts.pos_tag == "VERB"][0:10]
verbs
```

```python
# most common verbs
adj= pos_df_counts[pos_df_counts.pos_tag=='ADJ'][:10]
adj
```

```python
# extract the tokens and entity tags into a dataframe
ner_df = pd.DataFrame(columns=['token', 'ner_tag'])

for token in spacy_doc.ents:
    if pd.isna(token.label_) is False:
        ner_df = pd.concat([ner_df, pd.DataFrame.from_records(
            [{'token': token.text, 'ner_tag': token.label_}])], ignore_index=True)

ner_df.head()
```
